```python
import numpy as np

# Input: luas rumah (m²)
X = np.array([50, 60, 70, 80, 90], dtype=np.float32)

# Target: harga rumah (juta rupiah)
Y = np.array([250, 300, 350, 400, 450], dtype=np.float32)

print("X:", X)
print("Y:", Y)
```

```python
weight = 1.0
bias = 0.0

prediction = weight * X + bias

print("Prediction:", prediction)
```

```python
error = prediction - Y
squared_error = error ** 2
loss = np.mean(squared_error)

print("Loss:", loss)
```

```python
gradient_weight = 2 * np.mean((prediction - Y) * X)
gradient_bias   = 2 * np.mean(prediction - Y)
learning_rate = 0.0001
```

```python

weight = weight - learning_rate * gradient_weight
bias = bias - learning_rate * gradient_bias
```

```python
print("Loss:", loss)
print("Gradient weight:", gradient_weight)
print("Gradient bias:", gradient_bias)
print("Weight baru:", weight)
print("Bias baru:", bias)
```

```python
prediction = weight * X + bias

new_loss = np.mean((prediction - Y) ** 2)

print("Prediction baru:", prediction)
print("Loss baru:", new_loss)
```

```python
weight = 1
bias = 0

for epoch in range(100):

    # 1. Prediction
    prediction = weight * X + bias

    # 2. Error
    error = prediction - Y
 
    # 3. Loss
    loss = np.mean(error ** 2)

    # 4. Gradient
    gradient_weight = 2 * np.mean(error * X)
    gradient_bias = 2 * np.mean(error)

    # 5. Update
    weight = weight - learning_rate * gradient_weight
    bias = bias - learning_rate * gradient_bias
```

```python
print("Weight:", weight)
print("Bias:", bias)
print("Loss:", loss)
```
